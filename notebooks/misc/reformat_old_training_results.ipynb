The A (baseline) and C (leave-one-out) experiment results were generated with an older states of this repo (commit 55c7ec72e57133f448b8e278f448ca41e9e02ad4), which we no longer support. These notebooks produced them:

https://github.com/JustinCheung168/domain-adaptation-ct/blob/55c7ec72e57133f448b8e278f448ca41e9e02ad4/A_pipeline.ipynb

https://github.com/JustinCheung168/domain-adaptation-ct/blob/55c7ec72e57133f448b8e278f448ca41e9e02ad4/C_pipeline.ipynb

This notebook adapts those results to be compatible with our reorganized codebase

```python
import os
import pandas as pd

def convert_csvs_using_dataset_column(input_folder: str, output_dir: str):
    """
    Converts multiple CSV files in a folder into structured directories based on the 'Dataset' column inside each file.
    Each file's rows are split so that each dataset row goes into its own folder as 'test_metrics.csv'.

    Args:
        input_folder: Path to the `results` folder containing input CSV files.
        output_dir: Base output directory where subfolders will be created.
    """

    # Dictionary for capitalization-sensitive folder names
    legacy_dataset_name_map = {
        "original": "original",
        "uniform_noise": "Uniform_Noise",
        "rotate_90deg": "Rotated_90deg",
        "ring_data": "ring_data",
    }

    # Old names for experiments
    legacy_expt_name_map = {
        "A1": "baseline_train_on_original",
        "A2": "baseline_train_on_ring_data",
        "A3": "baseline_train_on_Rotated_90deg",
        "A4": "baseline_train_on_Uniform_Noise",
        "C11": "augmentation_train_exclude_ring_data",
        "C10": "augmentation_train_exclude_Rotated_90deg",
    }

    os.makedirs(output_dir, exist_ok=True)

    for filename in os.listdir(input_folder):
        if filename.endswith(".csv"):
            file_path = os.path.join(input_folder, filename)

            expt_short_name = legacy_expt_name_map[os.path.basename(file_path).split()[0].split("test_results_")[1].split("_")[0]]
            fold_name = "fold" + os.path.basename(file_path).split()[0].split("fold_")[-1].split(".csv")[0]

            # Read the CSV
            df = pd.read_csv(file_path)

            # Process each row individually
            for _, row in df.iterrows():
                dataset_raw = str(row["Dataset"]).strip()
                folder_name = legacy_dataset_name_map[dataset_raw]

                dataset_dir = os.path.join(output_dir, expt_short_name, folder_name, fold_name)
                os.makedirs(dataset_dir, exist_ok=True)

                # Prepare single-row DataFrame
                single_row_df = pd.DataFrame([row])

                # Write
                output_file = os.path.join(dataset_dir, "test_metrics.csv")
                single_row_df.to_csv(output_file, index=False)
                print(f"Extracted legacy result to {output_file}")

    print(f"All files processed and saved under {output_dir}")

convert_csvs_using_dataset_column(
    input_folder = "/repo/results/metrics/legacy_results/results/",
    output_dir = "/repo/results/metrics/final_results/",
)
```
