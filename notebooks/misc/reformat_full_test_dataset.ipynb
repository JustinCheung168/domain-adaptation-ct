Turn unnormalized `DG_Test_set/full_test_dataset.npz` containing all domains' data into many normalized npz files (one for each test domain) which are ready to provide to `run_training.py`. 

```python
import os

from domain_adaptation_ct.dataset.image_dataset import TwoLabelDataset
from domain_adaptation_ct.preprocess.organamnist_preprocessing import normalize_image

import numpy as np

full_test_dataset = np.load("/data/DG_Test_set/full_test_dataset.npz")
output_dir = "/data/DG_Test_set/"

print(full_test_dataset.keys())
label = full_test_dataset['label']
for dataset_name in full_test_dataset.keys():
    if 'label' not in dataset_name:
        TwoLabelDataset(
            images = normalize_image(full_test_dataset[dataset_name]),
            labels1 = label,
            labels2 = np.ones_like(label), # Doesn't matter
            convert_grayscale_to_rgb = True,
        ).save(os.path.join(output_dir, "test_dataset_"+dataset_name+".npz"))
```
